# Preparação do dashboard e dos filtros

## Dashboard
Dashboard Frigus — Visão de transações

## Fonte prevista
workspace.frigus_bi.transacoes_dashboard

## Dataset
Transacoes Frigus

## Filtro 1 — Período
- Campo: data_transacao
- Tipo: Date range picker
- Permite selecionar data inicial e final

## Filtro 2 — Método de pagamento
- Campo: metodo_pagamento
- Tipo: Single value com opção All, ou Multiple values

## Escopo
Todas as visualizações da página devem usar o mesmo dataset
e responder aos dois filtros

```python
%sql
SELECT
    transacao_id,
    data_transacao,
    metodo_pagamento,
    status_transacao,
    valor_transacao,
    tentativas,
    teve_retry,
    faixa_valor
FROM workspace.frigus_bi.transacoes_dashboard
```

## Visualizações previstas

| Visualização | Configuração |
|---|---|
| Card: total de transações | COUNT(transacao_id) |
| Card: valor total | SUM(valor_transacao) |
| Card: valor médio | AVG(valor_transacao) |
| Linha: evolução diária | Data no eixo X; contagem no eixo Y |
| Barras: métodos de pagamento | Método e contagem de transações |
| Barras: status | Status e contagem de transações |
| Distribuição por faixa de valor | Faixa e contagem de transações |

Todas usarão o dataset Transacoes Frigus

A soma dos valores será apresentada como valor das transações.
O cálculo de receita aprovada depende da confirmação dos status
que representam pagamentos concluídos
